# Methodes de descente pour l'optimisation sans contrainte

Ce notebook presente et compare quatre algorithmes qui cherchent le minimum d'une fonction f de R^n dans R :

1. Descente du gradient a pas fixe
2. Descente du gradient a pas variable (recherche lineaire d'Armijo)
3. Descente du gradient a pas conjugue (gradient conjugue)
4. Methode de quasi-Newton (BFGS)

Deux fonctions de test sont utilisees :

- une fonction quadratique en dimension 2 (cas ideal, solution connue),
- la fonction de Rosenbrock (non convexe, vallee etroite, cas difficile).

Plan du notebook :

1. Outils communs
2. Les quatre algorithmes
3. Fonctions de test
4. Experience 1 : fonction quadratique
5. Experience 2 : fonction de Rosenbrock
6. Tableau recapitulatif et conclusion

## 1. Outils communs

```python
import time
import numpy as np
import matplotlib.pyplot as plt

# Reglages d'affichage des graphiques
plt.rcParams["figure.figsize"] = (7, 5)
plt.rcParams["axes.grid"] = True

# Affichage plus lisible des nombres
np.set_printoptions(precision=6, suppress=True)
```

### Historique

Chaque algorithme enregistre, a chaque iteration, la position x, la valeur f(x)
et la norme du gradient. Ces informations servent ensuite aux graphiques.

```python
def nouvel_historique(x, f, grad):
    """Cree un historique contenant uniquement le point de depart."""
    return {"x": [x.copy()], "f": [f(x)], "g": [np.linalg.norm(grad(x))]}


def ajouter(historique, x, f, grad):
    """Ajoute l'etat courant (position, valeur, norme du gradient) a l'historique."""
    historique["x"].append(x.copy())
    historique["f"].append(f(x))
    historique["g"].append(np.linalg.norm(grad(x)))
```

### Recherche lineaire d'Armijo (backtracking)

Etant donne un point x et une direction de descente d, on cherche un pas alpha tel que :

f(x + alpha d) <= f(x) + c * alpha * grad_f(x)^T d

On part d'un pas initial alpha0 et on le multiplie par rho (0 < rho < 1)
tant que la condition n'est pas verifiee.
Cette fonction est utilisee par le pas variable, le gradient conjugue non lineaire et BFGS.

```python
def recherche_armijo(f, x, fx, g, d, alpha0=1.0, c=1e-4, rho=0.5, max_reductions=60):
    """
    Recherche lineaire par reduction successive du pas (backtracking).

    Parametres
    ----------
    f      : fonction a minimiser
    x      : point courant
    fx     : valeur de f en x (evite de la recalculer)
    g      : gradient de f en x
    d      : direction de descente (on doit avoir g @ d < 0)
    alpha0 : pas initial
    c      : constante de la condition d'Armijo (valeur faible, 1e-4 usuelle)
    rho    : facteur de reduction du pas
    """
    alpha = alpha0
    pente = g @ d  # derivee directionnelle, negative pour une direction de descente
    for _ in range(max_reductions):
        if f(x + alpha * d) <= fx + c * alpha * pente:
            return alpha
        alpha *= rho
    return alpha  # on renvoie le dernier pas teste si rien n'a fonctionne
```

## 2. Les quatre algorithmes

### 2.1 Descente du gradient a pas fixe

Iteration :

x_{k+1} = x_k - alpha * grad_f(x_k)

Le pas alpha est constant. Pour une fonction quadratique de plus petite valeur propre mu
et de plus grande valeur propre L (de la hessienne), la methode converge si 0 < alpha < 2/L.
Le meilleur pas est alpha = 2/(L + mu).
Un pas trop grand fait diverger la methode, un pas trop petit la rend tres lente.

```python
def descente_pas_fixe(f, grad, x0, alpha, tol=1e-6, max_iter=10000):
    """
    Descente du gradient a pas fixe.

    Parametres
    ----------
    f, grad  : fonction et son gradient
    x0       : point de depart
    alpha    : pas constant
    tol      : arret lorsque la norme du gradient est inferieure a tol
    max_iter : nombre maximal d'iterations
    """
    x = np.array(x0, dtype=float)
    historique = nouvel_historique(x, f, grad)

    for k in range(max_iter):
        g = grad(x)
        # Critere d'arret : gradient quasiment nul
        if np.linalg.norm(g) < tol:
            break
        # Mise a jour avec un pas constant
        x = x - alpha * g
        ajouter(historique, x, f, grad)

    return x, historique
```

### 2.2 Descente du gradient a pas variable

Iteration :

x_{k+1} = x_k - alpha_k * grad_f(x_k)

A chaque iteration, le pas alpha_k est calcule par la recherche d'Armijo.
La direction reste l'oppose du gradient, seul le pas s'adapte.

```python
def descente_pas_variable(f, grad, x0, tol=1e-6, max_iter=10000, alpha0=1.0):
    """
    Descente du gradient a pas variable (recherche lineaire d'Armijo).

    Le pas est recalcule a chaque iteration, ce qui evite de devoir
    connaitre a l'avance une bonne valeur de alpha.
    """
    x = np.array(x0, dtype=float)
    historique = nouvel_historique(x, f, grad)

    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            break
        d = -g                                   # direction de plus forte descente
        alpha = recherche_armijo(f, x, f(x), g, d, alpha0=alpha0)
        x = x + alpha * d
        ajouter(historique, x, f, grad)

    return x, historique
```

### 2.3 Descente du gradient a pas conjugue (gradient conjugue)

Principe : on remplace la direction -grad_f par une direction conjuguee aux precedentes.

d_0 = -grad_f(x_0)
x_{k+1} = x_k + alpha_k d_k
d_{k+1} = -grad_f(x_{k+1}) + beta_k d_k

Deux versions sont codees.

Version quadratique : f(x) = 1/2 x^T A x - b^T x avec A symetrique definie positive.
Le pas optimal est connu explicitement :

alpha_k = (r_k^T r_k) / (d_k^T A d_k), avec r_k = b - A x_k = -grad_f(x_k)
beta_k = (r_{k+1}^T r_{k+1}) / (r_k^T r_k)

En theorie, la methode converge en au plus n iterations (n = dimension).

Version non lineaire : pour une fonction quelconque, on utilise une recherche d'Armijo
pour le pas, et la formule de Fletcher-Reeves (FR) ou de Polak-Ribiere (PR+) pour beta :

beta_FR = (g_{k+1}^T g_{k+1}) / (g_k^T g_k)
beta_PR = max(0, g_{k+1}^T (g_{k+1} - g_k) / (g_k^T g_k))

```python
def gradient_conjugue_quadratique(A, b, x0, tol=1e-6, max_iter=None):
    """
    Gradient conjugue pour f(x) = 0.5 x^T A x - b^T x (A symetrique definie positive).
    Equivalent a resoudre le systeme lineaire A x = b.
    """
    x = np.array(x0, dtype=float)
    n = len(x)
    if max_iter is None:
        max_iter = n  # convergence theorique en n iterations

    # Fonction et gradient utilises uniquement pour l'historique
    f = lambda z: 0.5 * z @ A @ z - b @ z
    grad = lambda z: A @ z - b
    historique = nouvel_historique(x, f, grad)

    r = b - A @ x      # residu = oppose du gradient
    d = r.copy()       # premiere direction : plus forte descente

    for k in range(max_iter):
        if np.linalg.norm(r) < tol:
            break
        Ad = A @ d
        alpha = (r @ r) / (d @ Ad)         # pas optimal le long de d
        x = x + alpha * d
        r_nouveau = r - alpha * Ad         # mise a jour du residu sans recalculer A x
        beta = (r_nouveau @ r_nouveau) / (r @ r)
        d = r_nouveau + beta * d           # nouvelle direction conjuguee
        r = r_nouveau
        ajouter(historique, x, f, grad)

    return x, historique


def gradient_conjugue_non_lineaire(f, grad, x0, variante="PR", tol=1e-6, max_iter=10000):
    """
    Gradient conjugue pour une fonction quelconque.

    variante : "FR" (Fletcher-Reeves) ou "PR" (Polak-Ribiere avec beta >= 0)
    """
    x = np.array(x0, dtype=float)
    historique = nouvel_historique(x, f, grad)

    g = grad(x)
    d = -g

    for k in range(max_iter):
        if np.linalg.norm(g) < tol:
            break

        # Securite : si d n'est pas une direction de descente, on repart du gradient
        if g @ d >= 0:
            d = -g

        alpha = recherche_armijo(f, x, f(x), g, d)
        x = x + alpha * d
        g_nouveau = grad(x)

        # Calcul du coefficient beta
        if variante == "FR":
            beta = (g_nouveau @ g_nouveau) / (g @ g)
        else:
            beta = max(0.0, g_nouveau @ (g_nouveau - g) / (g @ g))

        d = -g_nouveau + beta * d
        g = g_nouveau
        ajouter(historique, x, f, grad)

    return x, historique
```

### 2.4 Methode de quasi-Newton (BFGS)

La methode de Newton utilise la hessienne : x_{k+1} = x_k - H^{-1} grad_f(x_k).
Calculer et inverser la hessienne est couteux. Les methodes de quasi-Newton
construisent a la place une approximation H_k de l'inverse de la hessienne,
mise a jour a chaque iteration uniquement avec des gradients.

Iteration :

d_k = -H_k grad_f(x_k)
x_{k+1} = x_k + alpha_k d_k          (alpha_k par recherche d'Armijo)

On pose s_k = x_{k+1} - x_k et y_k = grad_f(x_{k+1}) - grad_f(x_k), puis la formule BFGS :

H_{k+1} = (I - rho s y^T) H_k (I - rho y s^T) + rho s s^T, avec rho = 1 / (y^T s)

La mise a jour n'est faite que si y^T s > 0, ce qui garantit que H_k reste
symetrique definie positive (donc d_k reste une direction de descente).

```python
def quasi_newton_bfgs(f, grad, x0, tol=1e-6, max_iter=10000):
    """
    Methode de quasi-Newton BFGS avec recherche lineaire d'Armijo.

    H approche l'inverse de la hessienne. On la initialise a l'identite,
    la premiere iteration est donc une descente de gradient simple.
    """
    x = np.array(x0, dtype=float)
    n = len(x)
    H = np.eye(n)
    I = np.eye(n)
    historique = nouvel_historique(x, f, grad)

    g = grad(x)

    for k in range(max_iter):
        if np.linalg.norm(g) < tol:
            break

        d = -H @ g                                   # direction quasi-Newton
        alpha = recherche_armijo(f, x, f(x), g, d)   # pas par backtracking

        x_nouveau = x + alpha * d
        g_nouveau = grad(x_nouveau)

        s = x_nouveau - x                            # deplacement
        y = g_nouveau - g                            # variation du gradient
        sy = s @ y

        # Mise a jour BFGS seulement si la condition de courbure est verifiee
        if sy > 1e-10:
            rho = 1.0 / sy
            H = (I - rho * np.outer(s, y)) @ H @ (I - rho * np.outer(y, s)) \
                + rho * np.outer(s, s)

        x, g = x_nouveau, g_nouveau
        ajouter(historique, x, f, grad)

    return x, historique
```

## 3. Fonctions de test

### Fonction quadratique

f(x) = 1/2 x^T A x - b^T x, avec A = [[10, 3], [3, 2]] et b = (1, 2).

Les valeurs propres de A sont 1 et 11, le conditionnement vaut donc 11.
Le minimum est la solution de A x = b.

### Fonction de Rosenbrock

f(x, y) = (1 - x)^2 + 100 (y - x^2)^2

Le minimum est en (1, 1) avec f = 0. La fonction possede une vallee courbe
et etroite, ce qui rend la descente du gradient tres lente.

```python
# ----- Fonction quadratique -----
A = np.array([[10.0, 3.0],
              [3.0, 2.0]])
b = np.array([1.0, 2.0])

def f_quad(x):
    return 0.5 * x @ A @ x - b @ x

def grad_quad(x):
    return A @ x - b

x_etoile_quad = np.linalg.solve(A, b)   # solution exacte
f_etoile_quad = f_quad(x_etoile_quad)

valeurs_propres = np.linalg.eigvalsh(A)
mu, L = valeurs_propres[0], valeurs_propres[-1]
print("Valeurs propres de A :", valeurs_propres)
print("Pas fixe optimal 2/(L+mu) :", round(2 / (L + mu), 4))
print("Pas fixe maximal 2/L      :", round(2 / L, 4))
print("Solution exacte           :", x_etoile_quad)


# ----- Fonction de Rosenbrock -----
def f_rosen(x):
    return (1 - x[0]) ** 2 + 100 * (x[1] - x[0] ** 2) ** 2

def grad_rosen(x):
    return np.array([
        -2 * (1 - x[0]) - 400 * x[0] * (x[1] - x[0] ** 2),
        200 * (x[1] - x[0] ** 2),
    ])

x_etoile_rosen = np.array([1.0, 1.0])
f_etoile_rosen = 0.0
```

### Fonctions d'affichage et de lancement

```python
def lancer(nom, algo, **kwargs):
    """Execute un algorithme, mesure le temps et renvoie un dictionnaire de resultats."""
    debut = time.perf_counter()
    x, hist = algo(**kwargs)
    duree = time.perf_counter() - debut
    return {"nom": nom, "x": x, "hist": hist, "temps": duree,
            "iterations": len(hist["f"]) - 1}


def tracer_trajectoires(f, resultats, x_etoile, titre, niveaux=None, marge=0.5, nb_points_max=2000):
    """Trace les courbes de niveau de f et les trajectoires de chaque methode."""
    # Limites du graphique calculees a partir de toutes les trajectoires
    tous = np.vstack([np.array(r["hist"]["x"]) for r in resultats] + [x_etoile])
    xmin, ymin = tous.min(axis=0) - marge
    xmax, ymax = tous.max(axis=0) + marge

    X, Y = np.meshgrid(np.linspace(xmin, xmax, 300), np.linspace(ymin, ymax, 300))
    Z = np.array([[f(np.array([xi, yi])) for xi, yi in zip(ligne_x, ligne_y)]
                  for ligne_x, ligne_y in zip(X, Y)])

    plt.figure(figsize=(8, 6))
    plt.contour(X, Y, Z, levels=niveaux if niveaux is not None else 25, cmap="Greys", alpha=0.6)

    for r in resultats:
        traj = np.array(r["hist"]["x"])
        # On allege le trace si la trajectoire est tres longue
        pas = max(1, len(traj) // nb_points_max)
        traj = traj[::pas]
        plt.plot(traj[:, 0], traj[:, 1], "-o", markersize=3, label=r["nom"])

    plt.plot(*x_etoile, "k*", markersize=14, label="Minimum")
    plt.xlim(xmin, xmax)
    plt.ylim(ymin, ymax)
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.title(titre)
    plt.legend()
    plt.show()


def tracer_convergence(resultats, f_etoile, titre):
    """Trace l'erreur f(x_k) - f* et la norme du gradient en echelle logarithmique."""
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))

    for r in resultats:
        erreur = np.maximum(np.array(r["hist"]["f"]) - f_etoile, 1e-17)
        axes[0].semilogy(erreur, label=r["nom"])
        axes[1].semilogy(np.maximum(r["hist"]["g"], 1e-17), label=r["nom"])

    axes[0].set_title("Erreur sur la valeur : f(x_k) - f*")
    axes[1].set_title("Norme du gradient")
    for ax in axes:
        ax.set_xlabel("Iteration")
        ax.legend()
    fig.suptitle(titre)
    plt.show()


def afficher_tableau(resultats, x_etoile):
    """Affiche un tableau recapitulatif des resultats."""
    print(f"{'Methode':<34}{'Iterations':>11}{'Erreur sur x':>15}{'Norme gradient':>17}{'Temps (s)':>11}")
    print("-" * 88)
    for r in resultats:
        erreur_x = np.linalg.norm(r["x"] - x_etoile)
        print(f"{r['nom']:<34}{r['iterations']:>11d}{erreur_x:>15.2e}{r['hist']['g'][-1]:>17.2e}{r['temps']:>11.4f}")
```

## 4. Experience 1 : fonction quadratique

Point de depart : x0 = (-2, 3).
Pour la descente a pas fixe, on prend alpha = 0.1, valeur inferieure a 2/L = 0.18.

```python
x0_quad = np.array([-2.0, 3.0])

res_quad = [
    lancer("Gradient pas fixe (0.1)", descente_pas_fixe,
           f=f_quad, grad=grad_quad, x0=x0_quad, alpha=0.1),
    lancer("Gradient pas variable (Armijo)", descente_pas_variable,
           f=f_quad, grad=grad_quad, x0=x0_quad),
    lancer("Gradient conjugue", gradient_conjugue_quadratique,
           A=A, b=b, x0=x0_quad),
    lancer("Quasi-Newton BFGS", quasi_newton_bfgs,
           f=f_quad, grad=grad_quad, x0=x0_quad),
]

afficher_tableau(res_quad, x_etoile_quad)
```

```python
tracer_trajectoires(f_quad, res_quad, x_etoile_quad,
                    "Fonction quadratique : trajectoires")
```

```python
tracer_convergence(res_quad, f_etoile_quad, "Fonction quadratique : convergence")
```

Observations attendues :

- Le gradient conjugue atteint le minimum en 2 iterations (dimension 2), comme le predit la theorie.
- BFGS converge en tres peu d'iterations, sans calculer de hessienne.
- Les deux descentes de gradient avancent en zigzag et demandent beaucoup plus d'iterations.

### Influence du pas fixe

On compare plusieurs valeurs de alpha pour la descente a pas fixe :
un pas trop petit est lent, le pas optimal est le plus rapide, un pas superieur a 2/L fait diverger la methode.

```python
pas_a_tester = [0.02, 0.1, 2 / (L + mu), 0.19]
resultats_pas = []

for a in pas_a_tester:
    # max_iter limite pour eviter des valeurs gigantesques en cas de divergence
    r = lancer(f"alpha = {a:.4f}", descente_pas_fixe,
               f=f_quad, grad=grad_quad, x0=x0_quad, alpha=a, max_iter=300)
    resultats_pas.append(r)

plt.figure()
for r in resultats_pas:
    erreur = np.maximum(np.array(r["hist"]["f"]) - f_etoile_quad, 1e-17)
    plt.semilogy(erreur, label=r["nom"])
plt.xlabel("Iteration")
plt.ylabel("f(x_k) - f*")
plt.title("Descente a pas fixe : influence de alpha")
plt.legend()
plt.show()

print("Limite theorique de convergence : 2/L =", round(2 / L, 4))
```

## 5. Experience 2 : fonction de Rosenbrock

Point de depart classique : x0 = (-1.2, 1).
La descente a pas fixe demande un pas tres petit (alpha = 0.001) pour ne pas diverger,
et on lui autorise davantage d'iterations.

```python
x0_rosen = np.array([-1.2, 1.0])

res_rosen = [
    lancer("Gradient pas fixe (0.001)", descente_pas_fixe,
           f=f_rosen, grad=grad_rosen, x0=x0_rosen, alpha=0.001, max_iter=50000),
    lancer("Gradient pas variable (Armijo)", descente_pas_variable,
           f=f_rosen, grad=grad_rosen, x0=x0_rosen, max_iter=50000),
    lancer("Gradient conjugue (PR)", gradient_conjugue_non_lineaire,
           f=f_rosen, grad=grad_rosen, x0=x0_rosen, variante="PR", max_iter=50000),
    lancer("Gradient conjugue (FR)", gradient_conjugue_non_lineaire,
           f=f_rosen, grad=grad_rosen, x0=x0_rosen, variante="FR", max_iter=50000),
    lancer("Quasi-Newton BFGS", quasi_newton_bfgs,
           f=f_rosen, grad=grad_rosen, x0=x0_rosen, max_iter=50000),
]

afficher_tableau(res_rosen, x_etoile_rosen)
```

```python
niveaux_rosen = np.logspace(-1, 3.5, 20)
tracer_trajectoires(f_rosen, res_rosen, x_etoile_rosen,
                    "Rosenbrock : trajectoires", niveaux=niveaux_rosen, marge=0.3)
```

```python
tracer_convergence(res_rosen, f_etoile_rosen, "Rosenbrock : convergence")
```

Observations attendues :

- La descente a pas fixe est tres lente : le pas doit rester petit a cause de la forte courbure de la vallee.
- Le pas variable est plus robuste mais zigzague encore dans la vallee.
- Le gradient conjugue non lineaire fait mieux que les descentes simples.
- BFGS est de loin le plus efficace : l'approximation de la hessienne lui permet de suivre la vallee.

## 6. Conclusion

Comparaison des quatre methodes :

- Descente a pas fixe : la plus simple a coder, mais le choix de alpha est delicat
  (trop grand : divergence, trop petit : lenteur) et il depend de la fonction.
- Descente a pas variable : plus robuste car le pas s'adapte, au prix de quelques evaluations
  supplementaires de f a chaque iteration. La convergence reste lente sur les problemes mal conditionnes.
- Gradient conjugue : excellent sur les problemes quadratiques (convergence en n iterations au plus),
  tres faible besoin de memoire (quelques vecteurs). Sur les fonctions non quadratiques, sa performance
  depend de la formule de beta et de la qualite de la recherche lineaire.
- Quasi-Newton (BFGS) : convergence superlineaire, tres bon compromis entre rapidite et cout.
  Il demande de stocker une matrice n x n, ce qui devient couteux en grande dimension
  (on utilise alors la variante L-BFGS).
